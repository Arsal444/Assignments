# Python Loops & Conditional Statements

### Q1: 1 se 20 tak numbers print karein, lekin sirf EVEN (jift) numbers. (Hint: use % 2 == 0)

```python
numbers = int(input("Even Numbers"))
for n in range(1,21):
    if n % 2 == 0:
        print(n)
```

### Q2: 1 se 20 tak ODD numbers print karein using loop and if.

```python
numbers = int(input("Odd Numbers"))
for n in range(1,20):
    if n % 2 != 0:
        print(n)
```

### Q3: 1 se 10 tak loop chalayen. Agar number 5 ho tou 'Five found!' print karein, warna number print karein.

```python
for i in range(1,11):
    if i == 5:
        print("Five found!")
    else:
        print(i)
```

### Q4: Ek list banayen [3, 8, 12, 1, 9]. Loop laga kar sirf wo numbers print karein jo 5 se baray (greater) hon.

```python
numbers = [3, 8, 12, 1, 9]
        #  0  1   2  3  4
for i in numbers:
    if i > 5:
        print(i)
```

### Q5: 1 se 15 tak loop chalayen. Jo numbers 3 se divide hotay hain, unhe print karein.

```python
for n in range(1,16):
    if n % 3 == 0:
        print(n)
```

### Q6: List [2, -5, 10, -8, 0] par loop lagayen. Agar number > 0 hai tou 'Positive', < 0 hai tou 'Negative' aur 0 hai tou 'Zero' print karein.

```python
numbers = [2, -5, 10, -8, 0]

for n in numbers:
    if n > 0:
        print('Positive')
    elif n < 0:
        print('Negative')
    else:
        print('Zero')
```

### Q7: 1 se 10 tak loop chalayen, lekin jab number 7 aaye tou loop ko `break` kar dein.

```python
for i in range(1,11):
    if i == 7:
        break
    print(i)
```

### Q8: 1 se 10 tak loop chalayen, lekin number 4 ko `continue` (skip) kar dein.

```python
for n in range(1,11):
    if n == 4:
        continue
    print(n)
```

### Q9: String 'python' par loop lagayen. Agar character vowel (a, e, i, o, u) ho tou usay print karein.

```python
name = "python"
for vowel in name:
    if vowel in "a,e,i,o,u":
        print(vowel)
```

### Q10: Ages ki list [12, 18, 25, 15] lein. Agar age 18 ya us se zyada ho tou 'Adult' print karein, warna 'Minor'.

```python
ages = [12, 18, 25, 15]
for age in ages:
    if age >= 18:
        print('Adult')
    else:
        print('Minor')
```

### Q11: Names ki list ['Ali', 'Sara', 'Zain'] mein loop chalayen. Agar naam 'Sara' ho tou print karein 'Sara is present!'.

```python
names = ['Ali', 'Sara', 'Zain']
for name in names:
    if name == 'Sara':
        print('Sara is present!')        
```

### Q12: 1 se 20 tak numbers par loop lagayen. Jo number 3 aur 5 dono se divide ho (jaise 15), uspe 'FizzBuzz' print karein.

```python
for num in range(1,21):
    if num % 3 == 0 and num % 5 == 0:
        print(num, 'FizzBuzz')
```

### Q13: 1 se 10 tak numbers loop karein. Agar number even ho tou usko 2 se multiply kar ke print karein, warna waise hi print kar dein.

```python
for num in range(1,11):
    if num % 2 == 0:
        print('Even number', num*2)
    else:
        print('Odd number', num)
```

### Q14: List ['Ali', 'Bilal', 'Ahmed', 'Sana'] mein sirf wo naam print karein jo 'A' se shuru hotay hain.

```python
list = ['Ali', 'Bilal', 'Ahmed', 'Sana']
for name in list:
    if name.startswith('A'):
        print(name)
```

### Q15: Ek variable count = 0 banayen. List ['apple', 'mango', 'apple', 'banana', 'apple'] par loop lagayen aur count karein ke 'apple' kitni baar aya hai.

```python
count = 0
fruits = ['apple', 'mango', 'apple', 'banana', 'apple']
for fruit in fruits:
    if fruit == 'apple':
        count += 1
print(count, 'baar apple aaya hai')
```

### Q16: Marks ki list [45, 80, 30, 95, 40] par loop lagayen. Agar marks 50 se zyada hain tou 'Pass' print karein, warna 'Fail'.

```python
marks = [45, 80, 30, 95, 40]
for marks in marks:
    if marks >= 50:
        print('Pass')
    else:
        print('Fail')
```

### Q17: List [150, 45, 200, 99, 120] mein se sirf wo numbers print karein jo 100 aur 200 ke darmiyan hon.

```python
list = [150, 45, 200, 99, 120]
for num in list:
    if num > 100 and num < 200:
        print(num)
```

### Q18: 10 se 1 tak reverse loop chalayen. Jab number 5 aaye tou print karein 'Halfway there!'.

```python
for num in range(10, 0, -1):
    if num == 5:
        print('Halfway there!')
    else:
        print(num)
```

### Q19: 1 se 10 tak tamam EVEN numbers ka sum (jama) calculate karein ek variable use kar ke, aur aakhir mein print karein.

```python
total = 0
for num in range(1,11):
    if num % 2 == 0:
        total += num
print(total)
```

### Q20: Password check: List ['123', 'abc', 'admin123', 'xyz'] par loop lagayen. Agar 'admin123' mil jaye tou 'Access Granted' print kar ke loop break kar dein.

```python
password_check_list = ['123', 'abc', 'admin123', 'xyz']
for password in password_check_list:
    if password == 'admin123':
        print("Access Granted!")
        break
```
