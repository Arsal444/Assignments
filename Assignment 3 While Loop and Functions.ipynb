## Part 1: While Loop

### **Q1.** Ek while loop banayein jo 'Hello' ko 3 martaba print kare.

```python
i = 0
while i < 3:
    print('Hello')
    i += 1
```

### **Q2.** 10 se 1 tak ulti ginti (reverse counting) print karne ka while loop likhein.

```python
count = 10
while count >= 1:
    print(count)
    count -= 1
```

### **Q3.** Ek variable `i = 1` lein aur while loop ke zariye sirf odd (taaq) numbers print karein 10 tak.

```python
i = 1
print("odd numbers")
while i <= 10:
    print(i)
    i += 2
```

### **Q4.** While loop ka istemal kar ke 5 ka table sirf 5 tak print karein.

```python
t = 1
while t <= 5:
    print(5*t)
    t += 1
```

### **Q5.** Ek infinite loop banayein aur `break` statement use kar ke usay foran rok dein.

```python
while True:
    print('loop chal raha hai')
    break
```

### **Q6.** Ek while loop likhein jo `x = 5` se shuru ho, aur har dafa `x` mein se 1 minus kare jab tak 0 na ho.

```python
x = 5
while x >= 0:
    print(x)
    x -= 1
```

### **Q7.** Ek program likhein jo user se password mange, aur jab tak user "1234" na likhe, wo mangta rahe.

```python
password = input("Enter your password")
while password != "1234":
    password = input("Wrong password! Dobara enter karen:")
print("Correct password!")
```

### **Q8.** Ek while loop likhein jo `count = 0` se shuru ho aur har baar `count += 2` kare jab tak 10 na ho jaye.

```python
count = 0 
while count <= 10:
    print(count)
    count += 2
```

### **Q9.** String "Ali" ke har harf (letter) ko while loop aur index ka istemal karke alag line mein print karein.

```python
name = "Ali"
i = 0
while i < len(name):
    print(name[i])
    i += 1
```

### **Q10.** Ek while loop mein `continue` ka istemal karein taake wo number 3 ko skip kar ke 1 se 5 tak print kare.

```python
i = 1
while i <= 5:
    if i == 3:
        i += 1
        continue
    print(i)
    i += 1
```

## Part 2: Functions

### Q1. Ek function `greet(name)` banayein jo user ka naam argument mein le kar usay "Hello" kahe.

```python
def greet(name):
    print("Hello", name)
greet("asad")
```

### **Q2.** Ek function `add_numbers(a, b)` banayein jo do numbers ko add kar ke result return kare.

```python
def add_numbers(a,b:
    return a+b
result = add_numbers(2,4)
print(result)
```

### **Q3.** Ek function `check_even_odd(n)` banayein jo bataye ke number even hai ya odd.

```python
def check_even_odd(n):
    if n % 2 == 0:
        print("even num", n)
    else:
        print("odd num", n)
check_even_odd(2)
check_even_odd(5)
```

### **Q4.** Ek function `find_maximum(a, b, c)` banayein jo 3 numbers mein se sabse bada number bataye.

```python
def find_maximum(a,b,c):
    print(max(a,b,c))
find_maximum(12,55,81)
```

### **Q5.** Ek function `circle_area(radius)` banayein jo circle ka area calculate kare.

```python
def circle_area(radius):
    return 3.1415 * radius ** 2
print(circle_area(5))
```

### **Q6.** Ek function `celsius_to_fahrenheit(c)` banayein jo temperature convert kare.

```python
def celsius_to_fahrenheit(c):
    return (c*9/5) + 32
celsius_to_fahrenheit(25)
```

### **Q7.** Ek function `is_palindrome(word)` banayein jo check kare ke string palindrome hai ya nahi.

```python
def is_palindrome(word):
    if word == word[::-1]:
        print("is a palindrome", word)
    else:
        print("is not a palindrome", word)
is_palindrome("madam")
is_palindrome("sir")
is_palindrome("civic")
is_palindrome("corolla")
```

### **Q8.** Ek function `get_square(n)` banayein jo kisi bhi number ka square return kare.

```python
def get_square(n):
    return n**2
print(get_square(5))
```

### **Q9.** Ek function `calculate_factorial(n)` banayein jo number ka factorial return kare.

```python
def calculate_factorial(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result
print(calculate_factorial(5))
```

### **Q10.** Ek function `count_vowels(text)` banayein jo string mein vowels (a, e, i, o, u) ki tadad bataye.

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count
print(count_vowels("Hello World"))
print(count_vowels("Hello python"))
```
