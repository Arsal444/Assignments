### 6. Rectangle Area

### 7. Time Converter
Store 250 minutes in a variable. Calculate and print the number of hours and remaining minutes. 

```python
# 7. Time Converter 

minutes = 320

hours = minutes // 60
remaining_minutes = minutes % 60

print("Hours:", hours)
print("Remaining Minutes:", remaining_minutes)
```

### 8. Student Eligibility
Create variables for marks and attendance.
A student is eligible if:
- Marks are greater than or equal to 50
- Attendance is greater than or equal to 75
Use logical operators to calculate and print whether the student is eligible.

```python
# 8. Student Eligibility

marks = 65
attendance = 80

eligible = marks >= 50 and attendance >= 75

print("Marks:", marks)
print("Attendance:", attendance)
print("Student is eligible:", eligible)
```

### 9. Logical Operators 
Create two boolean variables:

print the result of:

```python
# 9. Logical Operators

a = True
b = False

print("a and b:", a and b)
print("a or b:", a or b)
print("not a:", not a)
print("not b:", not b)
```

# Final Challenge Student Report 
Create variable for:
 - Student name
 - Age
 - Marks of 3 subjects
 - Attendance
Then:
    1. Print the student's information 
    2. Calculate and print the total marks.
    3. Calculate and print the average marks.
    4. Use logical operators to calculate and print whether the student is eligible:
        Average marks>=50
        Attendance>=75

```python
# Final Challenge Student Report

student_name = "Ali"
age = 18
maths_marks = 75
english_marks = 65
science_marks = 80
attendance = 85

print("Student Name:", student_name)
print("Age:", age)
print("Maths Marks:", maths_marks)
print("English Marks:", english_marks)
print("Science Marks:", science_marks)
print("Attendance:", attendance, "%")

total_marks = maths_marks + english_marks + science_marks
print("Total Marks:", total_marks)

average_marks = total_marks / 3
print("Average Marks:", average_marks)

eligible = (average_marks >= 50) and (attendance >= 75)
print("Eligible:", eligible)
```
