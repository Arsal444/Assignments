### Q15

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
try:
    num = int(input("Enter a number"))
    result = 10 / num
    print(result)
except (ZeroDivisionError, ValueError):
    print("ZeroDivisionError ya ValueError aayi hai")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try: 
    a = int(input("Enter a number"))
except ValueError:
    print("Enter a valid number")
else:
    print(f"you entered {a}")
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    print("Age is valid")
try:
    age = int(input("Enter your age: "))
    check_age(age)
except ValueError as e:
    print(e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
try:
    file = open("abc.txt", "r")
    print(file.read())
except FileNotFoundError:
    print("File nahi mili")
finally:
    try:
        file.close()
        print("File handle closed")
    except NameError:
        pass
```
